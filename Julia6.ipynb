```julia
#Tässä notebookissa käsitellään vanhaa Savonian opinnäytetyökyselyä. Tarkoituksena on selvittää, että eroaako tyytyväisyys ohjaukseen 
#eri koulutusaloilla. 

#ladataan aluksi data Excelistä dataframeen. Argumentti infer_eltypesin avulla
#yritetään päätellä sarakkeiden tietotyypit automaattisesti. 

using DataFrames, XLSX

data = DataFrame(XLSX.readtable("C:\\Users\\Installer\\Desktop\\opinnäytetyökysely.xlsx","Kysely", infer_eltypes=true))
```

```julia
#Selvitetään seuraavaksi vastaajien määrä sukupuolittain eri koulutusaloilla. Tehdään tämä ristiintaulukoinnin avulla. 
#Asenna ensin paketti Freqtables. 

using FreqTables

ristiin = freqtable(data, :Opiskeluala, :Sukupuoli)
```

```julia
#Tehdään ristiintaulukointi vielä toisinpäin. 

ristiin2 = freqtable(data, :Sukupuoli, :Opiskeluala)
```

```julia
#Tyytyväisyyttä ohjaukseen kysytään useassa kysymyksessä (Excelin sarakkeet H,N,O,Q-U). Lähestymisnäkökulmasta
#riippuen myös muita väittämiä voi ottaa mukaan, mutta käytetään näitä. 
#Tehdään valituista väittämistä ns. summamuuttuja, eli lasketaan vastaajakohtainen keskiarvo näistä.
using Statistics 

ohjaus_cols = [
    "Sain riittävästi ohjausta",
    "Ohjaajani panos tuki työtäni",
    "Saamani ohjaus oli asiantuntevaa",
    "Saamani ohjaus oli motivoivaa",
    "Työni ohjaaja vastasi nopeasti tiedusteluihini",
    "Ohjaustilanteet eivät tuntuneet minusta pelottavilta",
    "Ohjaajaani oli helppo lähestyä",
    "Luotin ohjaajani neuvoihin",
]

transform!(
    data,
    AsTable(ohjaus_cols) => ByRow(row -> mean(collect(row))) => :Kokonaistyytyväisyys,
)


```

```julia
#Tarkastellaan nyt kokonaistyytyväisyyden keskiarvoa koulutusaloittain. 

ristiin3 = combine(groupby(data, :Opiskeluala),
                   :Kokonaistyytyväisyys => mean => "Tyytyväisyys ohjaukseen")
```

```julia
#Näyttää siltä, että varsinkin tekniikan alan opiskelijat ovat tyytyväisempiä ohjaukseen kuin liiketalouden alan opiskelijat. 
#Tutkitaan seuraavaksi, että onko se todella näin. Ts. otetaan selville, että onko ristiintaulukoinnissa havaittu ero
#tilastollisesti merkittävä vai johtuuko se sattumasta. 

#Erotetaan liiketalouden ja tekniikan opiskelijat omiksi vektoreikseen. Otetaan mukaan ainoastaan kokonaistyytyväisyyden sarake.
#Tämä selkeyttää analyysia. 

tekniikka = data[data.Opiskeluala .== "Tekniikka", :Kokonaistyytyväisyys]
liiketalous = data[data.Opiskeluala .== "Liiketalous", "Kokonaistyytyväisyys"]

println(liiketalous)
```

```julia
#Tutkitaan seuraavaksi, että ovatko lasketut keskiarvot normaalisti jakautuneita. Normaalisuus on edellytys t-testien käyttämiselle. 
#Asenna ensin tarvittaessa kirjasto HypothesisTest.

#Käytetään testaamiseen Shapiron-Wilkin testiä, jota voidaan käyttää, kun havaintoja on vähän. Sitä voidaan käyttää
#noin 2000 havaintomäärään asti. Muodostetaan hypoteesit: H0: Aineisto on normaalisti jakautunut. 
#H1: Aineisto ei ole normaalisti jakautunut. 


using HypothesisTests

ShapiroWilkTest(liiketalous)
```

```julia
ShapiroWilkTest(tekniikka)
```

```julia
#Nyt molemmissa tapauksissa p-arvo on todella pieni ja testi antaa 95 % luottamuksella suositukseksi hylätä H0.
#Kumpikaan datoista ei siis ole normaalisti jakautunut, eikä t-testejä voida käyttää. Käydään kuitenkin seuraavaksi lävitse, 
#miten edettäisiin, jos data olisivat normaalisti jakautuneita. 

#Testataan seuraavaksi Levenen testillä ovatko varianssit yhtä suuret, jotta tiedetään mitä t-testiä käytetään. 

#Levenen testissä käytettävät hypoteesit: H0: Varianssit ovat yhtä suuret. H1: Varianssit eivät ole yhtä suuret. 

LeveneTest(liiketalous, tekniikka)
```

```julia
#Testin perusteella nollahypoteesia ei voi hylätä (p = 0,534) ja varianssit voidaan olettaa yhtä suuriksi. 
#Tehdään hypoteesit. 

#H0: Jakaumat ovat samanlaiset, eli liiketalouden ja tekniikan opiskelijoiden tyytyväisyydessä ei ole eroa. 
#H1: Jakaumat ovat erilaiset, eli liiketalouden ja tekniikan opiskelijoiden tyytyväisyydessä on eroa.

#Tehdään kahden riippumattoman otoksen t-testi. 

EqualVarianceTTest(liiketalous, tekniikka)



#Mikäli varianssit eivät olisi olleet yhtä suuria, olisimme voineet käyttää Welchin t-testiä, joka ei edellytä
#yhtäsuuria variansseja. 

#UnequalVarianceTTest(liiketalous, tekniikka)
```

```julia
#Nyt p-arvo on 0,0025 ja 95% varmuudella voimme hylätä H0:n. Tyytyväisyydessä ohjaukseen on siis tilastollisesti
#merkittävä ero. Testi tehtiin kuitenkin kaksisuuntaisena, joten ero voisi olla molempiin suuntiin. 

#Keskiarvosta kuitenkin nähdään, että tekniikan opiskelijat ovat tyytyväisempiä ohjaukseen, joten tämä voidaan ottaa
#johtopäätökseksi. Olisi myös mahdollista tehdä ns. yksisuuntainen t-testi, mutta se sivuutetaan. 

#Koska datat eivät olleet normaalisti jakautuneita, käytetään Mann-Whitneyn U-testiä, joka ei edellytä 
#normaalisuutta. Sitä voidaan käyttää myös normaalisti jakautuneelle datalle, mutta t-testit ovat parempi valinta, jos käyttöedellytykset
#täyttyvät. 

#H0: Jakaumat ovat samanlaiset, eli liiketalouden ja tekniikan opiskelijoiden tyytyväisyydessä ei ole eroa. 
#H1: Jakaumat ovat erilaiset, eli liiketalouden ja tekniikan opiskelijoiden tyytyväisyydessä on eroa.

MannWhitneyUTest(liiketalous, tekniikka)
```

```julia
#Testin tulosten perusteella päädytään samaan johtopäätökseen kuin t-testiä käytettäessä. Huomataan, että vaikka t-testin
#käyttöedellytykset eivät toteutuneet, niin siitä huolimatta se johti oikeaan lopputulokseen. Näin ei kuitenkaan aina ole. 

#HUOM! Mikäli datat olisivat olleet normaalisti jakautuneita ja varianssit yhtä suuria, niin Mann-Whitneyä ei olisi tarvinnut tehdä. 

```
